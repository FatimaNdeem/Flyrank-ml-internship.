## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is **Refresh / Content Opportunity Scoring**, which is a **scoring** task. The goal is to assign each webpage a score indicating how strongly it should be considered for a content refresh. Pages with higher scores can then be prioritized by the SEO or content team for review and updating.

```python
print("ML Task Type: Scoring")
print("Dataset Shape:", df.shape)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

2. Target or proxy

The target is a content opportunity score that ranks webpages for refresh review. The starter dataset does not contain a direct observed outcome showing whether a refresh would succeed, so this score is a defined proxy, constructed from observable signals such as recent performance, impressions, trend, content age, freshness, CTR, and engagement. It should be treated as a prioritization score, not as a prediction that a refresh will cause recovery.

```python
df.columns
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

3. Success metric

The main success metric is Precision@50. It measures how many of the top 50 webpages ranked for review meet the defined opportunity criteria. This matches the real decision because the content team has limited capacity and needs the most useful pages near the top of the queue. A higher Precision@50 means the ranking puts more suitable review candidates in the first 50 positions.

```python
print("Chosen Metric: Precision@50")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is **one webpage**. Each row in the dataframe represents one anonymized webpage together with its search performance and content-related features.

```python
print("Dataset Shape:", df.shape)
df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "refresh pages with low traffic" would ignore many important factors that influence webpage performance. Features such as impressions, clicks, search volume, and other content signals interact in complex ways. Machine learning can learn these patterns from historical data and help rank or score webpages for refresh, providing better decision support than a fixed if-else rule.
